## Veri bilimi
</div>

Colab ile popüler Python kitaplıklarının tüm avantajlarından yararlanarak veri analiz edip görselleştirebilirsiniz. Aşağıdaki kod hücresi rastgele veri oluşturmak için <strong>numpy</strong>'yi, bu veriyi görselleştirmek için de <strong>matplotlib</strong>'i kullanır. Kodu düzenlemek için hücreyi tıklamanız yeterlidir. Sonrasında düzenlemeye başlayabilirsiniz.

Kendi verilerinizi Google Drive hesabınızdan &#40;e-tablolar dahil&#41;, GitHub'dan ve diğer pek çok kaynaktan Colab not defterlerine aktarabilirsiniz. Veri içe aktarma ve Colab'in veri bilimi için nasıl kullanılabileceği hakkında daha fazla bilgi edinmek için <a href="#working-with-data">Verilerle Çalışma</a> bölümünün altındaki bağlantılara bakabilirsiniz.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import os
import random
import matplotlib.pyplot as plt
import cv2
import kagglehub
from glob import glob

# 1. VERİ SETİNİ UYANDIRMA (Yoksa İndirir, Varsa Yerini Gösterir)
print("Veri seti kontrol ediliyor...")
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
train_path = os.path.join(path, "chest_xray", "train")

# 2. GÖRÜNTÜLERİ LİSTELEME
normal_images = glob(os.path.join(train_path, 'NORMAL', '*.jpeg'))
pneumonia_images = glob(os.path.join(train_path, 'PNEUMONIA', '*.jpeg'))

if not normal_images:
    # Bazı Kaggle yapılarında klasör ismi alt alta binmiş olabilir, kontrol ediyoruz:
    train_path = os.path.join(path, "chest_xray", "chest_xray", "train")
    normal_images = glob(os.path.join(train_path, 'NORMAL', '*.jpeg'))
    pneumonia_images = glob(os.path.join(train_path, 'PNEUMONIA', '*.jpeg'))

# 3. RASTGELE ÖRNEKLER SEÇME
random_normal = random.sample(normal_images, 5)
random_pneumonia = random.sample(pneumonia_images, 5)

# 4. KLİNİK DENETİM PANELİ (Görselleştime)
plt.figure(figsize=(25, 12))

# Üst Sıra: NORMAL
for i, img_path in enumerate(random_normal):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.subplot(2, 5, i + 1)
    plt.imshow(img)
    plt.title(f"LABEL: NORMAL\n(Verify as MD)", color='green', fontsize=12)
    plt.axis('off')

# Alt Sıra: PNÖMONİ (Konsolidasyon/Opasite kontrolü yapacağın yer)
for i, img_path in enumerate(random_pneumonia):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.subplot(2, 5, i + 6)
    plt.imshow(img)
    plt.title(f"LABEL: PNEUMONIA\n(Verify as MD)", color='red', fontsize=12)
    plt.axis('off')

plt.tight_layout()
plt.show()
print(f"Toplam {len(normal_images)} Normal, {len(pneumonia_images)} Pnömoni görüntüsü bulundu.")
```

```
import os
import random
import numpy as np
import matplotlib.pyplot as plt
import cv2
import kagglehub
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from sklearn.utils import class_weight
from glob import glob

# --- A. VERİ SETİNİ HAZIRLAMA (256x256 Çözünürlük) ---
print("Veri seti hazırlanıyor...")
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
train_path = os.path.join(path, "chest_xray", "train")
test_path = os.path.join(path, "chest_xray", "test")

# Eğer klasör yapısı iç içe geçmişse düzeltiyoruz
if not os.path.exists(os.path.join(train_path, 'NORMAL')):
    train_path = os.path.join(path, "chest_xray", "chest_xray", "train")
    test_path = os.path.join(path, "chest_xray", "chest_xray", "test")

IMG_SIZE = (256, 256)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(rescale=1./255, rotation_range=10, zoom_range=0.1)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    train_path, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode='binary'
)

# --- B. SINIF AĞIRLIĞI (Bias Filtresi) HESABI ---
labels = train_generator.classes
weights = class_weight.compute_class_weight('balanced', classes=np.unique(labels), y=labels)
class_weights_dict = {i : weights[i] for i in range(len(weights))}
print(f"Sınıf Ağırlıkları Uygulandı: {class_weights_dict}")

# --- C. MODEL v2.0 MİMARİSİ (Dual-Validation Architecture) ---
model_v2 = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(256, 256, 3)),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(128, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

model_v2.compile(optimizer=Adam(1e-4), loss='binary_crossentropy', metrics=['accuracy', 'Recall'])

# --- D. EĞİTİM (Training) ---
print("\nEğitim Başlıyor... (Bu süreç 15-20 dakika sürebilir)")
history_v2 = model_v2.fit(
    train_generator,
    epochs=10,
    validation_data=test_datagen.flow_from_directory(
        test_path, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode='binary'
    ),
    class_weight=class_weights_dict
)

# --- E. MODELİ KAYDETME ---
model_v2.save('pnomoni_modelim_v2.h5')
print("\nv2.0 Başarıyla Eğitildi ve Kaydedildi!")
```

```
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
import seaborn as sns

# 1. Tahminleri Alıyoruz (Validation Set üzerinde)
print("Model performans analizi yapılıyor...")
test_gen = test_datagen.flow_from_directory(
    test_path, target_size=(256, 256), batch_size=32, class_mode='binary', shuffle=False
)
Y_pred = model_v2.predict(test_gen)
y_pred = (Y_pred > 0.5).astype(int)
y_true = test_gen.classes

# 2. Confusion Matrix Oluşturma
cm = confusion_matrix(y_true, y_pred)

# 3. Görselleştirme (Isı Haritası)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['NORMAL', 'PNEUMONIA'],
            yticklabels=['NORMAL', 'PNEUMONIA'])
plt.title('v2.0 Model: Klinik Başarı Karnesi (Confusion Matrix)')
plt.ylabel('Gerçek Etiket (Hekim Tanısı)')
plt.xlabel('Yapay Zeka Tahmini')
plt.show()

# 4. Detaylı Rapor (Precision, Recall, F1)
print("\n--- Klinik Metrik Raporu ---")
print(classification_report(y_true, y_pred, target_names=['NORMAL', 'PNEUMONIA']))

# 5. AUC-ROC Eğrisi (Modelin Ayırt Edicilik Gücü)
fpr, tpr, thresholds = roc_curve(y_true, Y_pred)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC Curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.title('Model Ayırt Edicilik Gücü (ROC Curve)')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.show()
```
