````python
import os

base_dir = "/kaggle/input/chest-xray-pneumonia/chest_xray"  # senin path farklıysa değiştir

print(os.listdir(base_dir))
````

````python
train_dir = base_dir + "/train"
print(os.listdir(train_dir))
````

````python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(150,150),
    batch_size=32,
    class_mode='binary'
)

test_generator = test_datagen.flow_from_directory(
    base_dir + "/test",
    target_size=(150,150),
    batch_size=32,
    class_mode='binary'
)
````

````python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(150,150,3)),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
````

````python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
````

````python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
````

````python
history = model.fit(
    train_generator,
    steps_per_epoch=100,
    epochs=5,
    validation_data=test_generator,
    validation_steps=50
)
````

````python
print(history.history.keys())
````

````python
import numpy as np
from tensorflow.keras.preprocessing import image

def tahmin_et(img_path):
    img = image.load_img(img_path, target_size=(150,150))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array /= 255.

    prediction = model.predict(img_array)
    prob = prediction[0][0]

    print(f"Görüntü: {img_path}")
    print(f"Pneumonia olasılığı: {prob:.2f}")

    if prob > 0.5:
        print("Sonuç: PNEUMONIA")
    else:
        print("Sonuç: NORMAL")

    print("-"*30)
````

````python
import matplotlib.pyplot as plt
import cv2

def screenshot_save(img_path, save_path):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img)
    plt.axis('off')

    # Tahmini al
    img_array = image.load_img(img_path, target_size=(150,150))
    img_array = image.img_to_array(img_array)
    img_array = np.expand_dims(img_array, axis=0)
    img_array /= 255.
    pred = model.predict(img_array)[0][0]

    label = "PNEUMONIA" if pred > 0.5 else "NORMAL"
    plt.title(f"{label} ({pred:.2f})")

    # Kaydet
    plt.savefig(save_path)
    plt.close()
````

````python
import os

# 'screenshots' dizinini oluştur (eğer yoksa)
os.makedirs('screenshots', exist_ok=True)

# NORMAL örnekleri
for i in range(2):
    screenshot_save(normal_path + "/" + os.listdir(normal_path)[i], f"screenshots/normal_{i}.png")

# PNEUMONIA örnekleri
for i in range(2):
    screenshot_save(pneumonia_path + "/" + os.listdir(pneumonia_path)[i], f"screenshots/pneumonia_{i}.png")
````
